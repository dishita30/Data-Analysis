```python
import numpy as np
import pandas as pd  
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv("googleplaystore.csv")
```

```python
impute = SimpleImputer(missing_values=np.nan,strategy='mean')
impute.fit(df.iloc[ : , 2 : 3].values)
df.iloc[: , 2 : 3] = impute.transform(df.iloc[: , 2:3].values)
df = df.dropna()
```

```python
df.isnull().sum()
```

```python
# HOW MANY APPS for free(WITHOUT BREAK)

c = 0
for i in df:
    if(i[1]=='ART_AND_DESIGN' and i[6]=='Free'):
        c+=1

print("There are",c,"free apps from ART_AND_DESIGN")

```

```python
# HOW MANY APPS for paid(WITHOUT BREAK)

c = 0
for i in df:
    if(i[1]=='ART_AND_DESIGN' and i[6]=='Paid'):
        c+=1

print("There are",c,"paid apps from ART_AND_DESIGN")
```

### How many apps are there in ART_AND_DESIGN with rating more than 4.5?

```python
c = 0
for i in df:
    if(i[1] == 'ART_AND_DESIGN' and i[2] > 4.5):
        c+=1
print("There are",c,"ART_AND_DESIGN apps with rating above 4.5")
```

### How many apps are there in FAMILY with rating more than 4.5 AND free?


```python
c = 0
for i in df:
    if(i[1] == 'FAMILY' and i[2] > 4.5 and i[6]=='Free'):
        c+=1
print("There are",c," free FAMILY apps with rating above 4.5")
```

### List All the free apps with rating more than 4.5 and category is FAMILY

```python
print("-"*100)
print("       These are the free apps with rating more than 4.5 in category FAMILY ")
print("-"*100)
for i in df:
    if(i[1] == 'FAMILY' and i[2] > 4.5 and i[6]=='Free'):
        print(i) #print all names
print("-"*100)
```
