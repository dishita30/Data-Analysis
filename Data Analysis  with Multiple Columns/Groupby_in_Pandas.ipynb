```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv("googleplaystore.csv")

impute = SimpleImputer(missing_values=np.nan , strategy="mean")
impute.fit(df.iloc[ :, 2 : 3].values)
df = df.dropna()
```

```python
df.head()
```

### Q1. Name the apps that are in ART_AND_DESIGN with rating more than 4.5 in ascending order WRT Ratings?

```python
# we have to sort the values with respect to RATINGS in ascending order

#preprocessing

df_pr = df[df["Category"]== "ART_AND_DESIGN"]
df_pr = df_pr[df_pr['Rating']> 4.5]
# sorting
df_sorted  = df_pr.sort_values(by = "Rating", ascending=True)
df_sorted

```

### For Descending we make Ascending = False

```python

df_pr = df[df["Category"]== "ART_AND_DESIGN"]
df_pr = df_pr[df_pr['Rating']> 4.5]
# sorting
df_des  = df_pr.sort_values(by = "Rating", ascending=False)
df_des
```
